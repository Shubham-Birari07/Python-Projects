```python
#Image gaussianblur,medianblur bluring


org_img = cv2.imread("images/example4.png")

# Apply blurs to original image
g = cv2.GaussianBlur(org_img, (5, 5), 0)  # Kernel size should be odd
m = cv2.medianBlur(org_img, 5)
b = cv2.bilateralFilter(org_img, 9, 75, 75)

# Resize all to same dimensions
target_size = (200, 200)
res_img = cv2.resize(org_img, target_size)
g = cv2.resize(g, target_size)
m = cv2.resize(m, target_size)
b = cv2.resize(b, target_size)

h = np.hstack((res_img, g, m, b))

cv2.imshow("jav blur", h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

# Read and resize the target image (where you want to find the object)
img = cv2.imread("images/example9.png")
img = cv2.resize(img, (500, 500))
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)  # Convert to HSV (3 channels)

# Read the reference image (the object you want to detect)
img1 = cv2.imread(r"C:\Users\Shubham Anil Birari\OneDrive\Pictures\Screenshots\Screenshot 2026-03-25 120559.png")
hsv2 = cv2.cvtColor(img1, cv2.COLOR_BGR2HSV)  # Convert to HSV (3 channels)

# Create 2D histogram for Hue and Saturation channels [0,1]
# from the reference image (hsv2)
hist = cv2.calcHist([hsv2], [0, 1], None, [180, 256], [0, 180, 0, 256])

# Normalize the histogram
cv2.normalize(hist, hist, 0, 255, cv2.NORM_MINMAX)

# Create back projection on the target image (hsv) using the histogram from reference
mask = cv2.calcBackProject([hsv], [0, 1], hist, [0, 180, 0, 256], 1)

# Apply morphological operations
ker = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, ker)
mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, ker)

# Apply threshold to create binary mask
_, thr = cv2.threshold(mask, 200, 255, cv2.THRESH_BINARY)

# Convert mask to 3-channel for bitwise operations
mask_3channel = cv2.merge([thr, thr, thr])

# Apply mask to original image
res = cv2.bitwise_and(img, mask_3channel)

# Display results
cv2.imshow("Result", res)
cv2.imshow("Original Image", img)
cv2.imshow("Mask", thr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Displaying  coordinates and color code of the points clicked on the image
def click_b(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        s = f"{x},{y}"
        cv2.putText(img,s,(x,y),cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,255),2)
        cv2.imshow("wscube tech",img)
    elif event == cv2.EVENT_RBUTTONDOWN:
        b = img[y,x,0]
        g = img[y,x,1]
        r = img[y,x,2]
        s = f"{b},{g},{r}"
        cv2.putText(img,s,(x,y),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),2)
        cv2.imshow("wscube tech",img)




img = cv2.imread("images/example5.png")

cv2.imshow("wscube tech",img)

cv2.setMouseCallback("wscube tech", click_b)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
